```python
import PyPDF2

# Open the PDF file
pdf_file = open("GCUdata.pdf", 'rb')  # Open in binary mode
pdf_reader = PyPDF2.PdfReader(pdf_file)

# Extract text from all pages (modify for specific pages if needed)
text = ""
for page_num in range(len(pdf_reader.pages)):
    page = pdf_reader.pages[page_num]  # Use square bracket notation
    text += page.extract_text()

pdf_file.close()  # Close the file
```

```python
print(text)
```

```python

text_splitter = RecursiveCharacterTextSplitter(
        
        chunk_size=1000,
        chunk_overlap=200,
        length_function=len
    )
chunks = text_splitter.split_text(text)

```

```python
from langchain.chains.question_answering import load_qa_chain

retriever = vector_store.as_retriever(score_threshold = 0.7)
def get_conversational_chain():
    # Initialize your ChatGroq model
    llm = ChatGroq(groq_api_key=groq_api_key, model_name="Llama3-8b-8192")
    
   
    
    # Define your prompt template with corrected input variables
    prompt_template = """
        Answer the question of user by using information from context, make sure to provide all the details, if the answer is not in
        context just say, "answer is not available ", don't provide the wrong answer\n\n
        CONTEXT: {context}
        
        QUESTION: {question}
    """
    
    # Initialize the prompt template with corrected input variables
    prompt = PromptTemplate(template=prompt_template, input_variables=["context", "question"])

    # Initialize ConversationChain with llm, prompt, memory, and any other necessary parameters
    chain = load_qa_chain(llm, chain_type="stuff", prompt=prompt)
    
    return chain

if __name__ == "__main__":
    # Get the conversational chain
    chain = get_conversational_chain()
```

```python
def user_input(user_question):
    embeddings = GoogleGenerativeAIEmbeddings(model = "models/embedding-001")

    new_db = FAISS.load_local("faiss_index7", embeddings)
    docs = new_db.similarity_search(user_question)

    chain = get_conversational_chain()

    response = chain({"input_documents": docs, "question": user_question}, return_only_outputs=True)

    print(response)  # Print the response
    print("Reply: ", response["output_text"])  # Print the output text

    
```
